# 5.outliers

```python
Q1 = df["study_hours_per_day"].quantile(0.25)
Q3 = df["study_hours_per_day"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["study_hours_per_day"] < lower) |
    (df["study_hours_per_day"] > upper)
]

outliers
print(outliers.shape)
display(outliers.value_counts().sum())
```

# 6.distribution of numerical features vs target

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.histplot(data=df, x="previous_exam_score", kde=True, ax=axes[0])
axes[0].set_title("Previous Exam Score Distribution")

sns.histplot(data=df, x="previous_gpa", kde=True, ax=axes[1])
axes[1].set_title("Previous GPA Distribution")

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.histplot(data=df, x="previous_gpa", kde=True, ax=axes[0])
axes[0].set_title("Previous GPA Distribution")

sns.boxplot(data=df, x="pass_status", y="previous_gpa", ax=axes[1])
axes[1].set_title("Previous GPA by Pass/Fail")

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.histplot(data=df, x="attendance_percentage", kde=True, ax=axes[0])
axes[0].set_title("Attendance Percentage Distribution")

sns.boxplot(data=df, x="pass_status", y="attendance_percentage", ax=axes[1])
axes[1].set_title("Attendance Percentage by Pass/Fail")

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.histplot(data=df, x="study_hours_per_day", kde=True, ax=axes[0])
axes[0].set_title("Study Hours per Day Distribution")

sns.histplot(data=df, x="previous_gpa", kde=True, ax=axes[1])
axes[1].set_title("Previous GPA Distribution")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.regplot(
    data=df,
    x="study_hours_per_day",
    y="previous_gpa",
    scatter=False,
    lowess=True,
    line_kws={"color": "steelblue"}
)
plt.xlabel("Study Hours per Day")
plt.ylabel("Previous GPA")
plt.title("Study Hours vs GPA (Trend)")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
plt.hexbin(df["previous_gpa"], df["study_hours_per_day"], gridsize=40, cmap="viridis", mincnt=1)
plt.colorbar(label="Count")
plt.xlabel("Previous GPA")
plt.ylabel("Study Hours per Day")
plt.title("GPA vs Study Hours (Density)")
plt.tight_layout()
plt.show()
```
